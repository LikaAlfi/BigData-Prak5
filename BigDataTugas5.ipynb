```python
#Tugas 1
from collections import defaultdict

# Dataset sederhana: (kota, suhu)
data = [
    ("Jakarta", 30),
    ("Bandung", 25),
    ("Surabaya", 33),
    ("Jakarta", 31),
    ("Bandung", 24),
    ("Surabaya", 32),
    ("Jakarta", 29),
    ("Bandung", 26)
]

# Tahap MAP
mapped = []
for kota, suhu in data:
    # key = kota, value = (suhu, 1) -> 1 sebagai counter
    mapped.append((kota, (suhu, 1)))

print("Hasil MAP:")
print(mapped)


# Tahap SHUFFLE & SORT
shuffled = defaultdict(list)
for key, value in mapped:
    shuffled[key].append(value)

print("\nHasil SHUFFLE & SORT:")
print(dict(shuffled))


# Tahap REDUCE
reduced = {}
for key, values in shuffled.items():
    total_suhu = sum(v[0] for v in values)
    total_count = sum(v[1] for v in values)
    rata_rata = total_suhu / total_count
    maksimum = max(v[0] for v in values)
    minimum = min(v[0] for v in values)

    reduced[key] = {
        "Rata-rata": round(rata_rata, 2),
        "Maksimum": maksimum,
        "Minimum": minimum
    }

print("\nHasil REDUCE (Statistik suhu tiap kota):")
for kota, hasil in reduced.items():
    print(f"{kota}: {hasil}")
```

```python
# Tugas 2
from collections import defaultdict

# Map Function
def map_function(text):
    for word in text.split():
        yield (word, 1)

# Reduce Function
def reduce_function(pairs):
    result = defaultdict(int)
    for word, count in pairs:
        result[word] += count
    return result

# Dataset berupa paragraf
paragraf = """
Halo, aku lika. Aku suka banget sama K-POP.
Aku suka banget sama NCT Dream.
Biasku di NCT Dream adalah Lee Jeno.
Tanggal 28 September besok mereka ada konser TSD 4 di Jakarta.
Aku berharap aku bisa nonton konser mereka besok.
"""

# Proses Map
mapped = list(map_function(paragraf))

print("Hasil Map (pairs):")
print(mapped)

# Proses Reduce
reduced = reduce_function(mapped)

print("\nHasil Reduce (Word Count):")
for word, count in reduced.items():
    print(f"{word}: {count}")
```

```python
# Tugas 5
from collections import defaultdict

# Map Function
def map_function(dataset):
    # dataset berupa list judul lagu yang diputar
    for lagu in dataset:
        yield (lagu, 1)  # setiap kali lagu diputar, hitung +1

# Reduce Function
def reduce_function(pairs):
    result = defaultdict(int)
    for lagu, count in pairs:
        result[lagu] += count
    return result

# Dataset: daftar lagu yang diputar (contoh data)
dataset_lagu = [
    "Smoothie - NCT Dream",
    "Candy - NCT Dream",
    "Sticker - NCT 127",
    "Hot Sauce - NCT Dream",
    "Sticker - NCT 127",
    "Hello Future - NCT Dream",
    "Poison - NCT Dream",
    "Hot Sauce - NCT Dream",
    "Hello Future - NCT Dream",
    "Sticker - NCT 127",
    "Candy - NCT Dream",
    "Graduation - NCT Dream",
    "Smoothie - NCT Dream",
    "Poison - NCT Dream"
]

# Proses Map
mapped = list(map_function(dataset_lagu))
print("Hasil Map (pairs):")
print(mapped)

# Proses Reduce
reduced = reduce_function(mapped)

print("\nHasil Reduce (Jumlah Lagu Diputar):")
for lagu, count in reduced.items():
    print(f"{lagu}: {count} kali")
```
